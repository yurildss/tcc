```python
import os
import shutil
import random
import re
from collections import defaultdict

def split_dataset_no_leakage(
    dataset_root,
    output_root,
    train_ratio=0.7,
    val_ratio=0.15,
    seed=42
):
    random.seed(seed)

    os.makedirs(output_root, exist_ok=True)

    for beat_class in os.listdir(dataset_root):
        class_path = os.path.join(dataset_root, beat_class)
        if not os.path.isdir(class_path):
            continue

        print(f"\n📦 Processando classe {beat_class}")

        # Agrupa arquivos por beat_id
        beat_groups = defaultdict(list)

        for file in os.listdir(class_path):
            match = re.match(rf"{beat_class}_beat_(\d+)_(?:original|aug_.*)\.csv",    file)
            if match:
                beat_id = match.group(1)
                beat_groups[beat_id].append(file)

        beat_ids = list(beat_groups.keys())
        random.shuffle(beat_ids)

        n_total = len(beat_ids)
        n_train = int(n_total * train_ratio)
        n_val = int(n_total * val_ratio)

        train_ids = beat_ids[:n_train]
        val_ids = beat_ids[n_train:n_train + n_val]
        test_ids = beat_ids[n_train + n_val:]

        splits = {
            "train": train_ids,
            "val": val_ids,
            "test": test_ids
        }

        for split_name, ids in splits.items():
            split_dir = os.path.join(output_root, split_name, beat_class)
            os.makedirs(split_dir, exist_ok=True)

            for beat_id in ids:
                for file in beat_groups[beat_id]:
                    src = os.path.join(class_path, file)
                    dst = os.path.join(split_dir, file)
                    shutil.copy(src, dst)

        print(f"  ✔ beats: {n_total}")
        print(f"    🟢 train: {len(train_ids)}")
        print(f"    🟡 val:   {len(val_ids)}")
        print(f"    🔵 test:  {len(test_ids)}")
```

```python
split_dataset_no_leakage(
    dataset_root="dataset",
    output_root="dataset_split_2",
    train_ratio=0.7,
    val_ratio=0.15,
    seed=42
)
```
